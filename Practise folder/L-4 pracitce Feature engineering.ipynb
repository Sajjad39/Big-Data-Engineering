## 2.3 One Hot Encoding

>Label encoding has the advantage that it is straightforward but it has the disadvantage that the numeric values can be “misinterpreted” by the algorithms. For example, the value of 0 is obviously less than the value of 4 but does that really correspond to the data set in real life? Does a wagon have “4X” more weight in our calculation than the convertible?

```python
pd.get_dummies(obj_df, columns=["drive_wheels"]).head()
```

>The pd.get_dummies() function is a pandas method used to perform one-hot encoding on categorical data. It converts categorical variables into a series of binary (0 or 1) columns, where each column represents a category and indicates whether or not that category is present in the original data

```python
pd.get_dummies(obj_df, columns=["body_style", "drive_wheels"], prefix=["body", "drive"]).head()
```

>One hot encoding, is very useful but it can cause the number of columns to expand greatly if there are many unique values in a column. For the number of values in this example, it is not a problem. However this gets really challenging to manage when there are many more options

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Load the dummy dataset
X_dummy = np.random.rand(100, 60)
y_dummy = np.random.randint(0, 2, 100)

# Split the dataset into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X_dummy, y_dummy, test_size=0.25, random_state=42)

# Create a logistic regression model
model = LogisticRegression()

# Fit the model to the training data
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred = model.predict_proba(X_test)[:, 1]

# Evaluate the model's performance
accuracy = np.mean(y_pred > 0.5 == y_test)
print("Accuracy:", accuracy)

# Identify the customers with the highest probability of churning
customers_at_risk_of_churn = X_test[y_pred > 0.5]

# Print the customers at risk of churning
print(customers_at_risk_of_churn)

```
